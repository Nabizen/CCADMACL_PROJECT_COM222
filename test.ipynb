```python
def sample_cluster(df, cluster_num, sample_size=15, random_state=41):
    """Filters reviews by cluster and review length, then samples a specified number of rows."""
    filtered_df = df[(df['cluster'] == cluster_num) & (df['review_text'].str.len() > 20)]
    return filtered_df.sample(min(len(filtered_df), sample_size), random_state=random_state)
```
